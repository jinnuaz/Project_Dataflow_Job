# Adidas Sales Data Pipeline - Apache Beam on Cloud Dataflow

This notebook demonstrates an Apache Beam pipeline that:
- Reads Adidas sales data from CSV files in GCS
- Validates and enriches records using side-input rules
- Writes clean data to BigQuery
- Routes invalid records to a dead-letter table

## Pipeline Architecture
```
GCS CSV → Parse/Validate/Enrich → BigQuery (clean)
                ↓
         BigQuery (dead-letter)
```

## 4. Constants & Helper Functions

````python
# Expected CSV headers for the Adidas sales dataset
EXPECTED_HEADERS = [
    "Retailer", "RetailerID", "InvoiceDate", "Region", "State", "City", "Product",
    "PricePerUnit", "UnitsSold", "TotalSales", "OperatingProfit", "OperatingMargin", "SalesMethod"
]

def utc_now_iso():
    """Returns the current UTC timestamp in ISO format."""
    return datetime.now(timezone.utc).isoformat()


def parse_rules(json_line: str) -> dict:
    """Parses a JSON line containing validation rules."""
    return json.loads(json_line)
````

## 5. ParseValidateEnrich DoFn

This is the core transformation that:
- Parses CSV lines into dictionaries
- Validates schema and values using side-input rules
- Enriches records with derived fields
- Routes invalid records to a dead-letter output

````python
class ParseValidateEnrich(beam.DoFn):
    """
    Parses CSV lines into dicts, validates schema & values using side-input rules,
    emits cleaned rows to BigQuery, and routes invalid rows to a dead-letter output.
    """

    def __init__(self):
        # Metrics for monitoring pipeline health
        self.total = Metrics.counter("adidas_demo", "rows_total")
        self.valid = Metrics.counter("adidas_demo", "rows_valid")
        self.invalid = Metrics.counter("adidas_demo", "rows_invalid")
        self.proc_ms = Metrics.distribution("adidas_demo", "per_record_processing_ms")

    def process(self, line: str, rules: dict):
        start = time.time()
        self.total.inc()

        try:
            # Parse CSV line
            row = next(csv.reader([line]))
            if len(row) != len(EXPECTED_HEADERS):
                raise ValueError(f"Expected {len(EXPECTED_HEADERS)} columns, got {len(row)}")

            rec = dict(zip(EXPECTED_HEADERS, row))

            # Validate required fields
            required = ["Retailer", "RetailerID", "InvoiceDate", "Region", "State", "Product", "SalesMethod"]
            for k in required:
                if not str(rec.get(k, "")).strip():
                    raise ValueError(f"Missing required field: {k}")

            # Type coercion
            retailer_id = int(rec["RetailerID"])
            price_per_unit = float(rec["PricePerUnit"])
            units_sold = int(rec["UnitsSold"])
            total_sales = float(rec["TotalSales"])
            operating_profit = float(rec["OperatingProfit"])
            operating_margin = float(rec["OperatingMargin"])

            sales_method = rec["SalesMethod"].strip()
            state = rec["State"].strip()
            region = rec["Region"].strip()

            # Validate SalesMethod against allowed values from rules
            allowed_methods = set(rules.get("allowed_sales_methods", []))
            if allowed_methods and sales_method not in allowed_methods:
                raise ValueError(f"Invalid SalesMethod '{sales_method}'. Allowed: {sorted(list(allowed_methods))}")

            # Validate State -> Region consistency
            region_by_state = rules.get("region_by_state", {})
            expected_region = region_by_state.get(state)
            if expected_region and region != expected_region:
                raise ValueError(f"Region mismatch for state '{state}': got '{region}', expected '{expected_region}'")

            # Operating margin sanity check
            if total_sales <= 0:
                raise ValueError("TotalSales <= 0, cannot validate OperatingMargin safely")

            derived_margin = (operating_profit / total_sales) * 100.0
            if abs(derived_margin - operating_margin) > 2.0:
                raise ValueError(f"OperatingMargin mismatch: got {operating_margin}, derived {derived_margin:.2f}")

            # Enrichment: calculate channel fee and net sales
            fee_rate = float(rules.get("channel_fee_rate", {}).get(sales_method, 0.0))
            net_sales = total_sales * (1.0 - fee_rate)

            # Build output record
            out = {
                "Retailer": rec["Retailer"].strip(),
                "RetailerID": retailer_id,
                "InvoiceDate": rec["InvoiceDate"].strip(),
                "Region": region,
                "State": state,
                "City": rec["City"].strip(),
                "Product": rec["Product"].strip(),
                "PricePerUnit": price_per_unit,
                "UnitsSold": units_sold,
                "TotalSales": total_sales,
                "OperatingProfit": operating_profit,
                "OperatingMargin": operating_margin,
                "SalesMethod": sales_method,
                "DerivedOperatingMargin": round(derived_margin, 2),
                "ChannelFeeRate": fee_rate,
                "NetSales": round(net_sales, 2),
                "ValidatedAt": utc_now_iso()
            }

            self.valid.inc()
            yield out

        except Exception as e:
            self.invalid.inc()
            yield beam.pvalue.TaggedOutput(
                "deadletter",
                {
                    "raw_line": line,
                    "error_message": str(e),
                    "ingested_at": utc_now_iso()
                }
            )
        finally:
            self.proc_ms.update(int((time.time() - start) * 1000))
````

## 6. BigQuery Schema Definitions

````python
# Schema for dead-letter table (invalid records)
DEADLETTER_SCHEMA = {
    "fields": [
        {"name": "raw_line", "type": "STRING", "mode": "REQUIRED"},
        {"name": "error_message", "type": "STRING", "mode": "REQUIRED"},
        {"name": "ingested_at", "type": "TIMESTAMP", "mode": "REQUIRED"}
    ]
}

# Clean table uses SCHEMA_AUTODETECT for flexibility
````

## 7. Pipeline Options

````python
def get_pipeline_options(runner: str = "DirectRunner") -> PipelineOptions:
    """
    Creates pipeline options based on the specified runner.
    
    Args:
        runner: 'DirectRunner' for local, 'DataflowRunner' for cloud
    
    Returns:
        Configured PipelineOptions
    """
    options_dict = {
        "runner": runner,
        "project": PROJECT_ID,
        "region": REGION,
        "temp_location": TEMP_LOCATION,
        "staging_location": STAGING_LOCATION,
        "save_main_session": True,
    }
    
    # Add Dataflow-specific options
    if runner == "DataflowRunner":
        options_dict.update({
            "job_name": f"adidas-sales-pipeline-{int(time.time())}",
            # Uncomment and configure as needed:
            # "num_workers": 2,
            # "max_num_workers": 10,
            # "machine_type": "n1-standard-2",
            # "disk_size_gb": 50,
        })
    
    pipeline_options = PipelineOptions.from_dictionary(options_dict)
    pipeline_options.view_as(SetupOptions).save_main_session = True
    
    return pipeline_options
````

## 8. Build and Run the Pipeline

````python
def run_pipeline(
    runner: str = "DirectRunner",
    input_csv: str = INPUT_CSV,
    rules_json: str = RULES_JSON,
    bq_clean_table: str = BQ_CLEAN_TABLE,
    bq_deadletter_table: str = BQ_DEADLETTER_TABLE
):
    """
    Builds and executes the Adidas sales data pipeline.
    
    Args:
        runner: Pipeline runner ('DirectRunner' or 'DataflowRunner')
        input_csv: GCS path to input CSV file(s)
        rules_json: GCS path to validation rules JSON
        bq_clean_table: BigQuery table for valid records
        bq_deadletter_table: BigQuery table for invalid records
    """
    print(f"Starting pipeline with {runner}...")
    print(f"  Input: {input_csv}")
    print(f"  Rules: {rules_json}")
    print(f"  Clean Table: {bq_clean_table}")
    print(f"  Dead-letter Table: {bq_deadletter_table}")
    print()
    
    pipeline_options = get_pipeline_options(runner)
    
    with beam.Pipeline(options=pipeline_options) as p:
        # Load validation rules as a side input
        rules_view = (
            p
            | "ReadRulesJson" >> beam.io.ReadFromText(rules_json)
            | "ParseRules" >> beam.Map(parse_rules)
            | "RulesAsList" >> beam.combiners.ToList()
            | "RulesSingleton" >> beam.Map(lambda xs: xs[0] if xs else {})
            | "RulesAsView" >> beam.transforms.util.View.AsSingleton()
        )

        # Read CSV data (skip header line)
        lines = (
            p
            | "ReadAdidasCsv" >> beam.io.ReadFromText(input_csv, skip_header_lines=1)
        )

        # Parse, validate, and enrich records
        parsed = (
            lines
            | "ParseValidateEnrich" >> beam.ParDo(ParseValidateEnrich(), rules=rules_view)
                .with_outputs("deadletter", main="valid")
        )

        valid_rows = parsed.valid
        dead_rows = parsed.deadletter

        # Write valid records to BigQuery
        _ = (
            valid_rows
            | "WriteCleanToBQ" >> WriteToBigQuery(
                table=bq_clean_table,
                schema=SCHEMA_AUTODETECT,
                create_disposition=BigQueryDisposition.CREATE_IF_NEEDED,
                write_disposition=BigQueryDisposition.WRITE_APPEND,
                method=WriteToBigQuery.Method.FILE_LOADS,
                custom_gcs_temp_location=TEMP_LOCATION
            )
        )

        # Write invalid records to dead-letter table
        _ = (
            dead_rows
            | "WriteDeadletterToBQ" >> WriteToBigQuery(
                table=bq_deadletter_table,
                schema=DEADLETTER_SCHEMA,
                create_disposition=BigQueryDisposition.CREATE_IF_NEEDED,
                write_disposition=BigQueryDisposition.WRITE_APPEND,
                method=WriteToBigQuery.Method.FILE_LOADS,
                custom_gcs_temp_location=TEMP_LOCATION
            )
        )
    
    print("\nPipeline completed successfully!")
````
